```python
#4-2
a=[71,67,73,61,79,59,83,87,72,79]
scores=[]
for score in a:
    scores.append(score)

final_scores=[]
for score in scores:
    rescore=score*0.8+20
    final_scores.append(rescore)

ave=sum(final_scores)/len(final_scores)
print(f'調整後の平均点は{ave}')
```

```python
#5-1
def judge_uruuyear(num):
    if num%2==0:    #偶数
        if num%400==0:    #400で割り切れるうるう年
            return True
        else:
            if num%100==0:
                return False
            else:
                return True
    else:    #奇数
        return False
```

```python
judge_uruuyear(2020)
```
