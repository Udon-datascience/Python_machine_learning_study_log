```python
'''
df.loc[index,columns] → index行、col列の取り出し


その他の使い方
・df.loc[条件,columns] → 条件を満たす行の、col列の取り出し
・df.loc[list,columns] → 複数行を参照したいとき、index名のlistを作り、col列の取り出し(colについても同様)
・df.loc[index] → index行、全列の取り出し

データフレーム内に、該当する行・列がないときは、新たに行列が追加される
例
x['RM2'] = x['RM']**2　　　　RM2という列ができる
x.loc[:, 'RM2'] = x['RM']**2
x.loc[2000] = [10, 7, 8, 100]　　　2000という列ができる

'''
```

```python
'''
pandasでのcolumns,indexの扱い

columns と index は行名・列名を表す重要な用語ですが、「データの選択」に使うのではなく、「行名・列名そのものを取得・設定するための属性や引数」として使われます。

x.columns → 列名を取得・設定
x.index → 行名を取得・設定
x.loc[行, 列] → 行・列を指定してデータを取得・代入
x['列名'] → 列を取得・追加



① columns を使う場面

列名を確認する
print(x.columns)
Index(['RM', 'LSTAT', 'AGE'], dtype='object')

列名を変更する
x.columns = ['A', 'B', 'C']




② index を使う場面

行番号（インデックス）を確認する
print(x.index)
RangeIndex(start=0, stop=506, step=1)

インデックスを変更する
x.index = ['a', 'b', 'c', ...]



③ データを取り出すときは .loc

例えば
x.loc['a']
はインデックスが 'a' の行
'''
```
