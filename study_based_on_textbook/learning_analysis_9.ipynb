```python
#6-4
pd.isnull(df)
df2=df.fillna(df.median())
```

```python
#6-5,6
df2.plot(kind='scatter',x='x0',y='target')
df2.plot(kind='scatter',x='x1',y='target')
df2.plot(kind='scatter',x='x2',y='target')
df2.plot(kind='scatter',x='x3',y='target')
```

```python
#6-7
no=df2[(df['x2']<-2)&(df['target']>100)].index
print(no)
df3=df2.drop(no,axis=0)
```

```python
print(df2.shape)
print(df3.shape)
```

```python
#6-8
col=df3.loc[:,:'x3']
x=col
t=df3['target']
```

```python
#6-9
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,t,test_size=0.2,random_state=1)
```

```python
#6-10
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
#6-11   平均絶対誤差→誤差の絶対値を求められる　　決定係数→汎用性が高い、わかりやすい　　　決定係数を使うべき

#6-12
model.score(x_test,y_test)    #めちゃくちゃ性能がいい
```
