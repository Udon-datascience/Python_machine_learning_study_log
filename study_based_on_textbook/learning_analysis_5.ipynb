#機械学習の勉強

```python
import pandas as pd
```

```python
#機械学習基礎文法
date=[ 
    [160,161,150,60],
    [162,170,169,180]
]

#データフレームの作成
df=pd.DataFrame(date,index=['ハピ','チャパ'],columns=['4月','5月','6月','7月'])
```

```python
df['5月']
```

```python
df[['4月','6月']]
```

```python
print(type(df['5月']))
print(type(df[['4月','6月']]))
```

```python
#決定木
from sklearn import tree
x_col=df[['4月','5月','6月']]
t_col=df[['7月']]

model=tree.DecisionTreeClassifier(random_state=0)
model.fit(x_col,t_col)
```

```python
chipi=[[160,140,130]]
chipi_df=pd.DataFrame(chipi,
                      columns=x_col.columns)
model.predict(chipi_df)
```

```python
dubi=[150,40,190]
daba=[30,199,200]
new_data=pd.DataFrame([dubi,daba],
                      columns=x_col.columns)
model.predict(new_data)
```

```python
model.score(x_col,t_col)
```
